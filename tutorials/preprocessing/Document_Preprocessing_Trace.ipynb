```python
import tempfile
from PIL import Image
from klygo import preprocessing as p, media, files
tutorial_workspace = tempfile.TemporaryDirectory()
root = files.path(tutorial_workspace.name)
image = Image.new("RGB", (64, 48), (120, 80, 40))
```

### preprocessing.get_trace

`preprocessing.get_trace(image: 'Any') -> 'List[Dict[str, Any]]'`

Return or update get trace using preprocessing geometry history.

- `image`: required input; see the signature for its type and default.

Returns and effects: trace data, restored geometry, or updated image. Image operations remain deferred when applied to a `media.LazyImage`; calling `materialize` decodes pixels.

**Basic use**

```python
import tempfile
from klygo import files
with tempfile.TemporaryDirectory() as temporary:
    source = media.save(files.path(temporary) / "input.png", image)
    lazy = media.load(source)[0]
    processed = p.resize((32, 24))(lazy)
    p.materialize(processed, output="pil")
    history = p.get_trace(processed)
    print(p.get_trace(processed))
```

**Practical variation**

```python
print(p.original_size(processed), p.processed_size(processed))
```

**Expected error and recovery**

```python
try:
    p.get_trace(unexpected_option=True)
except TypeError as error:
    print(type(error).__name__, "Check the public signature and retry")
```

### preprocessing.restore_boxes

## Put it together
