```python
import tempfile
from PIL import Image
from klygo import preprocessing as p, media, files
tutorial_workspace = tempfile.TemporaryDirectory()
root = files.path(tutorial_workspace.name)
image = Image.new("RGB", (64, 48), (120, 80, 40))
source = media.save(root / "lazy_input.png", image)
lazy_source = media.load(source)[0]
```

### preprocessing.crop

`preprocessing.crop(box: 'Sequence[int]') -> 'Operation'`

Build a crop image preprocessing operation.

- `box`: required input; see the signature for its type and default.

Returns and effects: Operation; `from_input_spec` returns Compose. Image operations remain deferred when applied to a `media.LazyImage`; calling `materialize` decodes pixels.

**Basic use**

```python
operation = p.crop((5, 5, 50, 40))
result = operation(image)
print(operation.name, type(result).__name__)
```

**Practical variation**

```python
staged = operation(lazy_source)
print(type(staged).__name__, staged.is_loaded)
```

**Expected error and recovery**

```python
try:
    p.crop(unexpected_option=True)
except TypeError as error:
    print(type(error).__name__, "Check the public signature and retry")
```
