```python
import tempfile
from PIL import Image
from klygo import preprocessing as p, media, files
tutorial_workspace = tempfile.TemporaryDirectory()
root = files.path(tutorial_workspace.name)
image = Image.new("RGB", (64, 48), (120, 80, 40))
```

### preprocessing.brightness_score

`preprocessing.brightness_score(preview_size: 'Tuple[int, int]' = (160, 160)) -> 'Metric'`

Measure or prepare the brightness score image statistic.

- `preview_size`: optional input; see the signature for its type and default.

Returns and effects: Metric or evaluated statistic. Image operations remain deferred when applied to a `media.LazyImage`; calling `materialize` decodes pixels.

**Basic use**

```python
print(p.brightness_score()(image))
```

**Practical variation**

```python
other = Image.new("RGB", image.size, "blue")
print(p.brightness_score()(other))
```

**Expected error and recovery**

```python
try:
    p.brightness_score(unexpected_option=True)
except TypeError as error:
    print(type(error).__name__, "Check the public signature and retry")
```
