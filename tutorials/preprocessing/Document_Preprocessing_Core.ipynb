### preprocessing.Operation

`preprocessing.Operation(name: 'str', params: 'Dict[str, Any]' = <factory>, backend: 'str' = 'auto', requires_pixels: 'bool' = True, changes_geometry: 'bool' = False, terminal: 'bool' = False) -> None`

Create or execute Operation for an image preprocessing pipeline.

- `name`: required input; see the signature for its type and default.
- `params`: optional input; see the signature for its type and default.
- `backend`: optional input; see the signature for its type and default.
- `requires_pixels`: optional input; see the signature for its type and default.
- `changes_geometry`: optional input; see the signature for its type and default.
- `terminal`: optional input; see the signature for its type and default.

Returns and effects: operation, pipeline, predicate, metric, or evaluated image value. Image operations remain deferred when applied to a `media.LazyImage`; calling `materialize` decodes pixels.

**Basic use**

```python
operation = p.Operation("identity", {"function": lambda value: value})
print(operation(image).size)
```

**Practical variation**

```python
print("Pipeline metadata:", p.inspect(p.resize((16, 16))))
```

**Expected error and recovery**

```python
try:
    p.Operation(unexpected_option=True)
except TypeError as error:
    print(type(error).__name__, "Check the public signature and retry")
```
