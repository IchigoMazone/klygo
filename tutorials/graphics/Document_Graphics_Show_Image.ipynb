### graphics.close

`graphics.close(*figures)`

- `*figures`: Matplotlib figure instances to close, or omit to close all open figures.

Returns `None`.

### graphics.show

`graphics.show(image, title="Image", width=None, backend="auto", figsize=(10, 10))`

- `image`: PIL image or RGB NumPy array.
- `title`: display title.
- `width`: optional positive output width in pixels.
- `backend`: `auto`, `matplotlib`, or another string for OpenCV.
- `figsize`: pyplot figure size in inches.

Returns `None` and displays the image. The Matplotlib examples use the noninteractive backend for repeatable execution.

```python
import tempfile
import matplotlib
matplotlib.use("Agg", force=True)
from klygo import graphics, files
from PIL import Image
import numpy as np
tutorial_workspace = tempfile.TemporaryDirectory()
root = files.path(tutorial_workspace.name)
```

```python
image = Image.new("RGB", (40, 20), "steelblue")
graphics.show(image, title="Sample", backend="matplotlib")
```

**Practical variation**

```python
graphics.show(np.full((20, 40, 3), 100, dtype=np.uint8), width=80, backend="matplotlib", figsize=(4, 2))
```

**Expected error and recovery**

```python
try:
    graphics.show(None, width=50, backend="matplotlib")
except AttributeError as error:
    print(type(error).__name__, "Pass a PIL image or NumPy array")
```

```python
# Close open figures without importing matplotlib.pyplot directly
fig = graphics.bars([{'strategy': 'demo', 'mAP50-95': 0.5}], show=False, return_figure=True)
graphics.close(fig)
print('Closed figure successfully')
```

## Put it together

```python
marked = graphics.draw(image, [[3, 3, 20, 15]], show_labels=False)
graphics.show(marked, title="Annotated", backend="matplotlib")
```
