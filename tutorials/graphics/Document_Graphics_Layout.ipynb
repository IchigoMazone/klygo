### graphics.cell

Configure a single cell for `graphics.compose()` and `graphics.grid()`.

```python
from PIL import Image
from klygo import graphics

c1 = graphics.cell(Image.new("RGB", (120, 80), (40, 120, 200)), title="Panel A")
c2 = graphics.cell(Image.new("RGB", (120, 80), (200, 90, 50)), title="Panel B")
```

### graphics.compose

Normalize heterogeneous cell sources (images, paths, `Detection`, `Figure`) into RGB images.

```python
normalized = graphics.compose([c1, c2])
print(len(normalized), normalized[0].source.size)
```

### graphics.grid

Arrange cells into a `rows x cols` publication-ready grid.

```python
fig = graphics.grid(normalized, rows=1, cols=2, title="Side-by-Side Grid", show=False, return_figure=True)
```

### graphics.save

Export any Matplotlib `Figure` to PNG, PDF, or SVG.

```python
from tempfile import TemporaryDirectory
from klygo import files

with TemporaryDirectory() as tmp:
    out = graphics.save(fig, files.path(tmp) / "grid.png", dpi=150)
    print("Saved:", out)
```

## Put it together

Combine `graphics.cell`, `graphics.compose`, `graphics.grid`, and `graphics.save` to assemble multi-panel figures cleanly.

## Remember

- Always pass `GridCell` objects created by `graphics.cell()` to `graphics.compose()`.
- Set `show=False, return_figure=True` when composing charts into a larger grid.
