### graphics.partitions

Plot image counts and split percentages returned by `reports.partition_summary()`.

```python
from klygo import graphics

splits = [{"split": "train", "images": 300, "ratio_pct": 60.0}, {"split": "val", "images": 200, "ratio_pct": 40.0}]
graphics.partitions(splits, show=False)
```

### graphics.curves

Plot training/validation loss or metric curves from `reports.aggregate_runs()`.

```python
grouped = {"groups": {"S1": {"runs": [], "history_mean": {"epoch": [1, 2], "train/box_loss": [1.4, 0.9], "mAP50-95": [0.4, 0.65]}, "history_std": {"train/box_loss": [0.05, 0.03], "mAP50-95": [0.01, 0.01]}}}}
graphics.curves(grouped, metric="box_loss", split="train", show=False)
graphics.curves(grouped, metric="mAP50-95", show=False)
```

### graphics.thresholds

Plot P/R/F1 curves or TP/FP stacked bars from `reports.threshold_audit()`.

```python
audit = {"best_threshold": 0.55, "rows": [{"threshold": 0.30, "tp": 94, "fp": 21, "precision": 0.82, "recall": 0.76, "f1": 0.79}, {"threshold": 0.55, "tp": 88, "fp": 5, "precision": 0.95, "recall": 0.72, "f1": 0.81}]}
graphics.thresholds(audit, kind="curves", show=False)
graphics.thresholds(audit, kind="counts", show=False)
```

### graphics.mining

Plot positive/background counts or confidence distributions from `reports.mining_breakdown()`.

```python
mining_rows = [{"strategy": "S3", "positive_images": 93, "background_images": 157, "boxes": 210, "mean_conf": 0.80, "median_conf": 0.82}, {"strategy": "S4", "positive_images": 250, "background_images": 0, "boxes": 544, "mean_conf": 0.85, "median_conf": 0.87}]
graphics.mining(mining_rows, kind="counts", show=False)
graphics.mining(mining_rows, kind="confidence", show=False)
```

### graphics.bars

Plot single-metric (± std) or multi-metric grouped benchmark bars.

```python
table = [{"strategy": "S1", "mAP50-95_mean": 0.653, "mAP50-95_std": 0.008}, {"strategy": "S4", "mAP50-95_mean": 0.684, "mAP50-95_std": 0.007}]
graphics.bars(table, metrics=("mAP50-95",), show=False)
```

## Put it together

Generate single-panel figures with `show=False, return_figure=True` and arrange them using `graphics.grid()`.

## Remember

- Every chart function accepts `show`, `save_path`, `dpi`, and `return_figure` for consistent scripting and publication export.
