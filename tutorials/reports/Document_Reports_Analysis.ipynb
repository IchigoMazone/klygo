### reports.aggregate_runs

Aggregate multi-seed runs into Mean ± Std tables, per-seed rows, and epoch curves.

```python
from klygo import reports, utils

run_a = {"config": {"seed": 42}, "best_epoch": 5, "total_time_s": 100.0, "best_metrics": {"mAP50-95": 0.65}, "history": {"epoch": [1], "train/box_loss": [1.0]}}
run_b = {"config": {"seed": 100}, "best_epoch": 5, "total_time_s": 102.0, "best_metrics": {"mAP50-95": 0.67}, "history": {"epoch": [1], "train/box_loss": [0.9]}}
agg = reports.aggregate_runs({"S1": [run_a, run_b]}, format="yolo")
utils.tabulate(agg["table"], title="Aggregated Runs")
```

### reports.threshold_audit

Sweep confidence thresholds, select optimal T*, and verify on a confirmation set.

```python
from klygo import outputs, reports

pred = outputs.Detections([outputs.Detection([outputs.Box((0, 0, 20, 20), label="hat", score=0.9)], source_id="a", image_size=(64, 64))])
target = outputs.Detections([outputs.Detection([outputs.Box((0, 0, 20, 20), label="hat")], source_id="a", image_size=(64, 64))])
audit = reports.threshold_audit(pred, target, thresholds=[0.3, 0.5, 0.8])
print("Best T*:", audit["best_threshold"])
```

### reports.mining_breakdown

Summarize positive/background images, box density, and confidence across mining subsets.

```python
rows = reports.mining_breakdown({"S4": pred})
utils.tabulate(rows, title="Mining Breakdown")
```
