### reports.load

Load YOLO training logs (`results.csv`, `args.yaml`) from a run directory.

```python
from tempfile import TemporaryDirectory
from klygo import files, reports

with TemporaryDirectory() as tmp:
    run = files.path(tmp) / "run"
    files.mkdir(run)
    files.save(run / "args.yaml", {"epochs": 2, "seed": 42}, overwrite=True, verbose=False)
    files.save(run / "results.csv", [{"epoch": 1, "metrics/mAP50-95(B)": 0.55, "time": 10.0}], overwrite=True, verbose=False)
    print(reports.load(run, format="yolo")["best_epoch"])
```

### reports.partition_summary

Summarize image, positive/background, and GT box counts across dataset partitions.

```python
from klygo import outputs, reports, utils

rows = reports.partition_summary({"train": outputs.Detections([]), "val": outputs.Detections([])}, format="yolo")
utils.tabulate(rows, title="Partition Summary")
```

## Put it together

Use `reports.load` and `reports.partition_summary` to feed `reports.aggregate_runs` and `graphics.partitions`.

## Remember

- Both `reports.load` and `reports.partition_summary` accept `format="yolo"`.
