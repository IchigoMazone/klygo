### models.load

`models.load(model: Union[str, Any], **kwargs) -> klygo.models.base.BaseModel`

Resolve a configured model or return an existing model wrapper.

- Parameters: model accepts a wrapper, loaded Config, metadata mapping, registry name, or supported model source; kwargs override declared groups. Read configuration files first with config.load(path) or Config(path).read().
- Returns and effects: see the public signature and examples.

**Basic use**

```python
from klygo import models
model = models.Detector()
print(models.load(model) is model)
```

**Practical variation**

```python
print(type(models.load(models.Detector(model_id="demo"))).__name__)
```

**Expected error and recovery**

```python
try:
    models.load('nonexistent-model-name')
except ValueError as error:
    print(type(error).__name__, 'Correct the input and retry')
```

### models.configure

`models.configure(model_id: str, metadata: Optional[Mapping[str, Any]] = None, flags: Union[Mapping[str, bool], Sequence[str], NoneType] = None, priority: Optional[Mapping[str, Any]] = None, details: Optional[Mapping[str, Any]] = None, implementation: Union[type, str, NoneType] = None, **kwargs) -> box.box.Box`

Bind an identifier and parameter values to metadata.

- Parameters: model_id identifies the model; metadata, flags, priority, details, and implementation define its schema; kwargs set group values.
- Returns and effects: see the public signature and examples.

**Basic use**

```python
from klygo import models
definition = models.configure("demo", flags=("model", "post"), model_device="cpu", post_threshold=0.5)
print(definition.model.device, definition.post.threshold)
```

**Practical variation**

```python
print(models.configure("demo", flags=("model",), model_device="cpu").flags)
```

**Expected error and recovery**

```python
try:
    models.configure('demo', flags=('model',), post_threshold=0.5)
except ValueError as error:
    print(type(error).__name__, 'Correct the input and retry')
```

### models.set_backend

`models.set_backend(backend: str, engine: Optional[str] = None) -> None`

Set the default backend through environment variables.

- Parameters: backend names the runtime; engine optionally sets the Keras engine.
- Returns and effects: see the public signature and examples.

**Basic use**

```python
from klygo import models
import os
from unittest.mock import patch
with patch.dict(os.environ, {}, clear=False):
    models.set_backend("custom")
    print(models.get_backend())
```

**Practical variation**

```python
import os
from unittest.mock import patch
with patch.dict(os.environ, {}, clear=False):
    models.set_backend("keras", engine="torch")
    print(os.environ["KERAS_BACKEND"])
```

**Expected error and recovery**

```python
try:
    models.set_backend()
except TypeError as error:
    print(type(error).__name__, 'Correct the input and retry')
```

### models.get_backend

`models.get_backend() -> str`

Read the selected backend or auto fallback.

- Parameters: No parameters.
- Returns and effects: see the public signature and examples.

**Basic use**

```python
from klygo import models
print(models.get_backend())
```

**Practical variation**

```python
import os
from unittest.mock import patch
with patch.dict(os.environ, {"KLYGO_BACKEND":"demo"}):
    print(models.get_backend())
```

**Expected error and recovery**

```python
try:
    models.get_backend('extra')
except TypeError as error:
    print(type(error).__name__, 'Correct the input and retry')
```
