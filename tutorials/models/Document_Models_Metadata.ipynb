### models.metadata

`models.metadata(flags: Union[Mapping[str, bool], Sequence[str]], priority: Optional[Mapping[str, Any]] = None, details: Optional[Mapping[str, Any]] = None) -> box.box.Box`

Combine flags, priority aliases, and descriptive model details.

- Parameters: flags declares groups; priority maps unprefixed parameter names; details describes the model.
- Returns and effects: see the public signature and examples.

**Basic use**

```python
from klygo import models
definition = models.metadata(models.flags("model"), details=models.details(name="Demo", task="Detection", backend="Custom", library="local"))
print(definition.flags, definition.details.name)
```

**Practical variation**

```python
print(models.metadata(models.flags("post")).keys())
```

**Expected error and recovery**

```python
try:
    models.metadata(models.flags('model'), details={})
except ValueError as error:
    print(type(error).__name__, 'Correct the input and retry')
```

### models.priority

`models.priority(flags: Union[Mapping[str, bool], Sequence[str]], **parameters: Any) -> box.box.Box`

Assign unprefixed parameter names to flagged groups.

- Parameters: flags declares group names; keyword arguments list priority names by destination group.
- Returns and effects: see the public signature and examples.

**Basic use**

```python
from klygo import models
groups = models.flags("model", "post")
print(models.priority(groups, model=("device",), post=("threshold",)))
```

**Practical variation**

```python
print(models.priority(models.flags("post"), post=("threshold",)).post)
```

**Expected error and recovery**

```python
try:
    models.priority(models.flags('model'), missing=('x',))
except ValueError as error:
    print(type(error).__name__, 'Correct the input and retry')
```

### models.details

`models.details(*, name: str, task: str, backend: str, library: str, **information: Any) -> box.box.Box`

Build validated human-readable model details.

- Parameters: name, task, backend, and library are required nonempty strings; extra keyword information is included unless None.
- Returns and effects: see the public signature and examples.

**Basic use**

```python
from klygo import models
info = models.details(name="Demo", task="Detection", backend="Custom", library="local", version="1")
print(info.name, info.version)
```

**Practical variation**

```python
print(models.details(name="Demo", task="Detection", backend="Custom", library="local", version=None))
```

**Expected error and recovery**

```python
try:
    models.details(name=' ', task='Detection', backend='Custom', library='local')
except ValueError as error:
    print(type(error).__name__, 'Correct the input and retry')
```

### models.flags

`models.flags(*groups: str, **permissions: bool) -> box.box.Box`

Declare parameter groups and predict-time override permissions.

- Parameters: Use positional group names or keyword bool permissions; do not mix forms.
- Returns and effects: see the public signature and examples.

**Basic use**

```python
from klygo import models
print(models.flags(model=False, post=True))
```

**Practical variation**

```python
print(models.flags("model", "post"))
```

**Expected error and recovery**

```python
try:
    models.flags('model', post=True)
except TypeError as error:
    print(type(error).__name__, 'Correct the input and retry')
```

## Put it together

```python
groups = models.flags("model", "post")
definition = models.metadata(groups, models.priority(groups, post=("threshold",)), models.details(name="Demo", task="Detection", backend="Custom", library="local"))
print(definition.details.name)
```
