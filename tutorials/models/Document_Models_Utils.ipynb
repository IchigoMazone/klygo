### models.utils

`models.utils (module)`

Access model metadata and warning helper functions.

- Parameters: Module namespace; helpers normalize flags and metadata.
- Returns and effects: see the public signature and examples.

**Basic use**

```python
from klygo import models
print(models.utils.normalize_flags(("model", "post")))
```

**Practical variation**

```python
print(models.utils.normalize_flags({"model":False,"post":True}))
```

**Expected error and recovery**

```python
try:
    models.utils.normalize_flags('model')
except TypeError as error:
    print(type(error).__name__, 'Correct the input and retry')
```

### models.errors

`models.errors (module)`

Access model-specific exception classes.

- Parameters: Module namespace containing model state, configuration, device, and dtype exceptions.
- Returns and effects: see the public signature and examples.

**Basic use**

```python
from klygo import models
try:
    raise models.errors.InvalidConfigError("bad setting")
except models.errors.ModelFrameworkError as error:
    print(type(error).__name__)
```

**Practical variation**

```python
print(issubclass(models.errors.InvalidStateError, models.errors.ModelFrameworkError))
```

**Expected error and recovery**

```python
try:
    raise models.errors.InvalidConfigError('bad setting')
except models.errors.InvalidConfigError as error:
    print(type(error).__name__, 'Correct the input and retry')
```

### models.suppress_warnings

`models.suppress_warnings(func=None)`

Temporarily silence Python warnings in a context or decorator.

- Parameters: func optionally decorates a callable; omitted func returns a context manager.
- Returns and effects: see the public signature and examples.

**Basic use**

```python
from klygo import models
import warnings
with models.suppress_warnings():
    warnings.warn("hidden", UserWarning)
print("warning scope exited")
```

**Practical variation**

```python
@models.suppress_warnings
def quiet():
    import warnings
    warnings.warn("hidden", UserWarning)
quiet()
```

**Expected error and recovery**

```python
try:
    with models.suppress_warnings():
            raise ValueError('still raised')
except ValueError as error:
    print(type(error).__name__, 'Correct the input and retry')
```

### models.suppress_ai_warnings

`models.suppress_ai_warnings() -> None`

Configure quiet logging for optional AI frameworks without importing them.

- Parameters: No parameters.
- Returns and effects: see the public signature and examples.

**Basic use**

```python
from klygo import models
models.suppress_ai_warnings()
print("logging configured")
```

**Practical variation**

```python
import logging
models.suppress_ai_warnings()
print(logging.getLogger("torch").level)
```

**Expected error and recovery**

```python
try:
    models.suppress_ai_warnings('extra')
except TypeError as error:
    print(type(error).__name__, 'Correct the input and retry')
```

## Put it together

```python
with models.suppress_warnings():
    models.suppress_ai_warnings()
print(models.utils.normalize_flags(("model",)))
```
