### models.base

`models.base (module)`

Access the abstract BaseModel implementation module.

- Parameters: Module namespace; BaseModel is abstract.
- Returns and effects: see the public signature and examples.

**Basic use**

```python
from klygo import models
print(models.base.BaseModel is models.BaseModel)
```

### models.BaseModel

`models.BaseModel(metadata: Optional[Dict[str, Any]] = None, backend: Optional[str] = None, model_id: Optional[str] = None, task: Optional[str] = None, **kwargs) -> None`

Define an abstract contract for model wrappers.

- Parameters: metadata, backend, model_id, and task initialize model identity and settings; subclasses implement abstract methods.
- Returns and effects: see the public signature and examples.

**Basic use**

```python
from klygo import models
print(issubclass(models.Detector, models.BaseModel))
```

**Practical variation**

```python
print(sorted(models.BaseModel.__abstractmethods__))
```

**Expected error and recovery**

```python
try:
    models.BaseModel()
except TypeError as error:
    print(type(error).__name__, 'Correct the input and retry')
```

### models.Detector

`models.Detector(metadata: Optional[Dict[str, Any]] = None, backend: Optional[str] = None, model_id: Optional[str] = None, **kwargs) -> None`

Build a detector wrapper for a concrete backend implementation.

- Parameters: metadata, backend, and model_id configure the wrapper; subclasses implement forward.
- Returns and effects: see the public signature and examples.

**Basic use**

```python
from klygo import models
detector = models.Detector()
print(detector.task, isinstance(detector, models.BaseModel))
```

**Practical variation**

```python
print(models.Detector(model_id="demo").model_id)
```

**Expected error and recovery**

```python
try:
    models.Detector().forward([])
except TypeError as error:
    print(type(error).__name__, "Pass both images and prompt")
```

## Put it together

```python
wrapper = models.Detector(model_id="demo")
print(isinstance(wrapper, models.BaseModel), models.load(wrapper) is wrapper)
```
