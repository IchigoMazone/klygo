## Try the public API

The setup cell resets only this tutorial's temporary workspace.

```python
from tempfile import TemporaryDirectory
from klygo import files, utils

if 'tutorial_workspace' in globals():
    tutorial_workspace.cleanup()
tutorial_workspace = TemporaryDirectory(prefix='klygo-utils-')
root = files.path(tutorial_workspace.name)
print(utils.human_size(1024 ** 2))
with utils.ProgressBar(2, 'Checking', verbose=False) as progress:
    for _ in range(2):
        progress.update()
print('Finished in', root)
```

## Remember

`human_size` uses steps of 1024. Use `ProgressBar` with `with` to close the renderer; `create_progress_bar` returns the same wrapper with different scaling defaults.
