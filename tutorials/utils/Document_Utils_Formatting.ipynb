## Jump to an API

- [utils.human_size](#utils.human_size)

### utils.human_size

`utils.human_size(num_bytes: int, decimal_places: int = 2) -> str`

- `num_bytes`: non-negative integer byte count.
- `decimal_places`: displayed precision; default `2`.
- Returns a string with B, KB, MB, GB, or TB using steps of 1024.

#### Basic use

```python
for count in (0, 512, 1024, 1024 ** 2):
    print(count, utils.human_size(count))
```

#### Practical variation

Display a local file's size with one decimal place.

```python
report = files.join(root, 'report.txt')
files.save(report, 'Klygo report', verbose=False)
print(utils.human_size(files.size(report), decimal_places=1))
```

#### Expected error and recovery

A negative byte count raises `ValueError`; a non-integer raises `TypeError`.

```python
for bad_value in (-1, 1.5):
    try:
        print(utils.human_size(bad_value))
    except (TypeError, ValueError) as error:
        print(type(error).__name__, error)
```

### utils.exclude_keys

`utils.exclude_keys(values, keys)` returns a shallow dictionary copy without one key or a list of keys. The source mapping remains unchanged.


```python
settings = {"threshold": 0.4, "dtype": "float32", "device": "cpu"}
model_settings = utils.exclude_keys(settings, ["dtype", "device"])
print(model_settings)
print(settings)
```

### utils.invert

`utils.invert(values)` exchanges unique, hashable mapping values with their keys and can be called again to restore the original mapping.


```python
classes = {"cat": 0, "dog": 1}
class_names = utils.invert(classes)
print(class_names)
print(utils.invert(class_names))
```

## Put it together

Format several Klygo-managed files without reimplementing byte units.

```python
for name, value in [('small.txt', 'a'), ('large.txt', 'x' * 2048)]:
    path = files.join(root, name)
    files.save(path, value, overwrite=True, verbose=False)
    print(files.name(path), utils.human_size(files.size(path)))
```
