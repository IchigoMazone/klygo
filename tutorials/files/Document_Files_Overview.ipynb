## Try the public API

The installation cell above fetches the `v2-release` branch. Examples below use a temporary folder; running setup again resets it.

```python
from tempfile import TemporaryDirectory
from klygo import files

# Run this cell again to replace only this notebook's temporary workspace.
if "tutorial_workspace" in globals():
    tutorial_workspace.cleanup()
tutorial_workspace = TemporaryDirectory(prefix="klygo-files-")
root = files.path(tutorial_workspace.name)
print("Example directory:", root)
```

```python
config_file = root / 'config.json'
files.save(config_file, {'batch': 16, 'classes': ['cat', 'dog']}, verbose=False)
print(files.load(config_file, verbose=False))
print(files.info(config_file)['size'], 'bytes')
```

```python
backup = files.copy(config_file, root / 'backup' / 'config.json')
print('Backup:', backup)
print('Same contents:', files.compare(config_file, backup))
```

## Remember

- Import the public API with `from klygo import files`.
- Path arguments accept strings or `pathlib.Path` (`~` is expanded by default).
- Most write operations protect an existing destination unless you pass `overwrite=True`.
- A returned path from `paths.py` is a pure transformation, not necessarily a created file on disk.
