## Jump to an API

- [validators.validate_type](#validators.validate_type)

### validators.validate_type

`validators.validate_type(value: object, expected: type | tuple[type, ...], name: str) -> None`

- `value`: the argument to check.
- `expected`: a type or tuple of accepted types, passed to `isinstance`.
- `name`: argument name in the error message.
- Returns `None` on success; raises `TypeError` on failure.

#### Basic use

```python
validators.validate_type('image.jpg', str, 'source')
print('A string path is accepted')
```

#### Practical variation

A path API may accept both strings and path-like values.

```python
import os
path = files.join(root, 'image.jpg')
validators.validate_type(path, (str, os.PathLike), 'source')
print('A pathlib path is accepted:', path)
```

#### Expected error and recovery

Catch a mismatched type and retry with a supported value.

```python
try:
    validators.validate_type(42, (str, os.PathLike), 'source')
except TypeError as error:
    print(error)
    validators.validate_type(str(42), (str, os.PathLike), 'source')
```

## Put it together

Validate a path before using the public `files` API.

```python
target = files.join(root, 'report.txt')
validators.validate_type(target, (str, os.PathLike), 'target')
files.save(target, 'validated', verbose=False)
print(files.load(target, verbose=False))
```
