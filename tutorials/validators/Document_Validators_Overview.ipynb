## Try the public API

Run setup again to reset this notebook's temporary workspace.

```python
from tempfile import TemporaryDirectory
from klygo import files, validators

if 'tutorial_workspace' in globals():
    tutorial_workspace.cleanup()
tutorial_workspace = TemporaryDirectory(prefix='klygo-validators-')
root = files.path(tutorial_workspace.name)
source = files.join(root, 'image.jpg')
validators.validate_type(source, (str, __import__('os').PathLike), 'source')
print('Accepted:', source)
validators.validate_count(4, "batch")
print("Validated count:", 4)
```

## Remember

`validate_type` returns `None` on success and raises `TypeError` with the argument name on failure. See the [type validator guide](Document_Validators_Type_Validator.ipynb).
