## Put it together


```python
from klygo import metrics, outputs

pred = outputs.Detections([outputs.Detection(
    [outputs.Box((0, 0, 20, 20), label="cat", score=0.9)],
    source_id="example", image_size=(100, 100),
)])
truth = outputs.Detections([outputs.Detection(
    [outputs.Box((0, 0, 20, 20), label="cat")],
    source_id="example", image_size=(100, 100),
)])

report = metrics.evaluate(pred, truth, verbose=True)
print(report)
```

## Remember

Progress defaults to `False`. `evaluate` owns one shared bar; `compose` only builds a reusable configuration. Progress counts CPU work and does not force GPU synchronization.
