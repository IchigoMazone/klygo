## Jump to an API

- `metrics.compose`
- `metrics.evaluate`


```python
from klygo import metrics, outputs

pred = outputs.Detections([outputs.Detection(
    [outputs.Box((0, 0, 20, 20), label="cat", score=0.9)],
    source_id="example", image_size=(100, 100),
)])
truth = outputs.Detections([outputs.Detection(
    [outputs.Box((0, 0, 20, 20), label="cat")],
    source_id="example", image_size=(100, 100),
)])
```

### metrics.compose

Build reusable configurations without running or displaying progress.


```python
evaluator = metrics.compose(metrics.precision_recall(), metrics.map_score())
```

### metrics.evaluate

Use one display across preparation, matching, and metric completion.


```python
report = metrics.evaluate(pred, truth, evaluator, verbose=True)
print(report)
```

## Put it together

Reuse the same evaluator for multiple materialized dataset pairs.

## Remember

Omit `verbose` for silent execution. Frame identities, image sizes, and devices must agree.
