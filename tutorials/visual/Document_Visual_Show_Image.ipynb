### visual.show_image

`visual.show_image(image, title="Image", width=None, backend="auto", figsize=(10, 10))`

- `image`: PIL image or RGB NumPy array.
- `title`: display title.
- `width`: optional positive output width in pixels.
- `backend`: `auto`, `matplotlib`, or another string for OpenCV.
- `figsize`: pyplot figure size in inches.

Returns `None` and displays the image. The Matplotlib examples use the noninteractive backend for repeatable execution.

```python
import tempfile
import matplotlib
matplotlib.use("Agg", force=True)
from klygo import visual, files
from PIL import Image
import numpy as np
tutorial_workspace = tempfile.TemporaryDirectory()
root = files.path(tutorial_workspace.name)
```

```python
image = Image.new("RGB", (40, 20), "steelblue")
visual.show_image(image, title="Sample", backend="matplotlib")
```

**Practical variation**

```python
visual.show_image(np.full((20, 40, 3), 100, dtype=np.uint8), width=80, backend="matplotlib", figsize=(4, 2))
```

**Expected error and recovery**

```python
try:
    visual.show_image(None, width=50, backend="matplotlib")
except AttributeError as error:
    print(type(error).__name__, "Pass a PIL image or NumPy array")
```

## Put it together

```python
marked = visual.draw_bboxes(image, [[3, 3, 20, 15]], show_labels=False)
visual.show_image(marked, title="Annotated", backend="matplotlib")
```
