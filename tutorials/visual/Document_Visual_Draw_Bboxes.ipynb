### visual.draw_bboxes

`visual.draw_bboxes(image, bboxes, labels=None, scores=None, thickness=None, line_width=None, font_size=None, box_color=None, text_color=None, show_labels=True, show_scores=True, **kwargs)`

- `image`: PIL image or RGB NumPy array.
- `bboxes`: pixel boxes as `[xmin, ymin, xmax, ymax]`.
- `labels`, `scores`: optional captions matched by position.
- `line_width`, `thickness`, `font_size`, `box_color`, `text_color`: rendering options.
- `show_labels`, `show_scores`: caption switches.

Returns an annotated copy in the same representation as the input.

```python
import tempfile
import matplotlib
matplotlib.use("Agg", force=True)
from klygo import visual, files
from PIL import Image
import numpy as np
tutorial_workspace = tempfile.TemporaryDirectory()
root = files.path(tutorial_workspace.name)
```

```python
image = np.full((60, 80, 3), 255, dtype=np.uint8)
annotated = visual.draw_bboxes(image, [[10, 10, 50, 40]], labels=["cat"], scores=[0.9])
print(annotated.shape, bool(np.all(image == 255)))
```

**Practical variation**

```python
plain = visual.draw_bboxes(Image.fromarray(image), [[10, 10, 50, 40]], box_color=(255, 0, 0), show_labels=False, show_scores=False)
print(plain.size)
```

**Expected error and recovery**

```python
try:
    visual.draw_bboxes(image, [[10, 20, 30]])
except ValueError as error:
    print(type(error).__name__, "A box needs four coordinates")
```

## Put it together

```python
canvas = Image.new("RGB", (80, 60), "white")
output = visual.draw_bboxes(canvas, [[4, 4, 36, 25], [40, 20, 76, 50]], labels=["first", "second"], line_width=2)
print(output.size)
```
