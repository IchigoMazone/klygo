### visual.plot_dataset_stats

`visual.plot_dataset_stats(source, figsize=(10, 6))`

- `source`: YOLO dataset directory or archive path.
- `figsize`: Matplotlib figure size in inches.

Returns `None` and plots object counts from matching image/label pairs.

**Basic use**

```python
import tempfile
import matplotlib
matplotlib.use("Agg", force=True)
from klygo import visual, files
from PIL import Image
import numpy as np
tutorial_workspace = tempfile.TemporaryDirectory()
root = files.path(tutorial_workspace.name)
```

```python
files.mkdir(root / "images")
files.mkdir(root / "labels")
files.save(root / "data.yaml", "names: [cat, dog]\nnc: 2\n", verbose=False)
Image.new("RGB", (8, 8)).save(root / "images" / "sample.png")
files.save(root / "labels" / "sample.txt", "0 0.5 0.5 0.2 0.2\n", verbose=False)
visual.plot_dataset_stats(root)
```

**Practical variation**

```python
visual.plot_dataset_stats(files.path(root), figsize=(6, 3))
```

**Expected error and recovery**

```python
try:
    visual.plot_dataset_stats(root / "missing")
except FileNotFoundError as error:
    print(type(error).__name__, "Create a dataset or pass an existing path")
```

## Put it together

```python
files.save(root / "labels" / "sample.txt", "0 0.5 0.5 0.2 0.2\n1 0.2 0.2 0.1 0.1\n", overwrite=True, verbose=False)
visual.plot_dataset_stats(root, figsize=(5, 3))
```
