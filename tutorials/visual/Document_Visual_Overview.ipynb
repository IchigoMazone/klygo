```python
import tempfile
import matplotlib
matplotlib.use("Agg", force=True)
from klygo import visual, files
from PIL import Image
import numpy as np
tutorial_workspace = tempfile.TemporaryDirectory()
root = files.path(tutorial_workspace.name)
```

```python
image = np.full((48, 72, 3), 255, dtype=np.uint8)
annotated = visual.draw_bboxes(image, [[5, 5, 35, 30]], labels=["item"])
print(annotated.shape, bool(np.array_equal(image, annotated)))
```
