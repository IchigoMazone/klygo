## Quick start

```python
from tempfile import TemporaryDirectory
from zipfile import BadZipFile
from klygo import archive, files

if "tutorial_workspace" in globals():
    tutorial_workspace.cleanup()
tutorial_workspace = TemporaryDirectory(prefix="klygo-archive-")
root = files.path(tutorial_workspace.name)
print("Example directory:", root)
```

```python
source = files.mkdir(root / "dataset")
files.save(source / "labels.txt", ["cat", "dog"], verbose=False)
files.save(source / "config.json", {"batch": 16, "classes": 2}, verbose=False)
images = files.mkdir(source / "images")
files.save(images / "cat.txt", "image placeholder", verbose=False)

sample_archive = root / "dataset.zip"
archive.compress(source, sample_archive, verbose=False)
print("Members:", archive.list_files(sample_archive))
```

```python
print("Format:", archive.detect_format(sample_archive))
print("Valid:", archive.test(sample_archive))
print("Information:", archive.get_info(sample_archive))

output = root / "extracted"
archive.extract(sample_archive, output, verbose=False)
print("Extracted:", [files.as_posix(files.relative(path, output)) for path in files.find(output)])
```

## Remember

- Import public APIs with `from klygo import archive, files` and build paths with `files.path()`.
- Archive writes protect existing destinations unless `overwrite=True` is explicit.
- ZIP and TAR-family archives use the Python standard library; 7-Zip (`py7zr`) and RAR (`rarfile`) use optional backends.
