```python
import tempfile
from PIL import Image
from klygo import outputs, files
tutorial_workspace = tempfile.TemporaryDirectory()
root = files.path(tutorial_workspace.name)
```

```python
image = Image.new("RGB", (80, 60), "white")
box = outputs.Box((10, 10, 40, 35), label="cat", score=0.9)
frame = outputs.Detection([box], source=image, source_id="overview", image_size=image.size)
results = outputs.Detections([frame])
print(results[0].image_size)
```

## Remember

A `Box` belongs to an image-level `Detection`; `Detections` groups frames. See [Visual overview](../graphics/Document_Graphics_Overview.ipynb), [Files overview](../files/Document_Files_Overview.ipynb), and [Archive overview](../archive/Document_Archive_Overview.ipynb).
