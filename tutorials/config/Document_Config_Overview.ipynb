## Try the public API

The setup cell below creates a temporary workspace with `files.path()`. Running it again resets the folder.

```python
from tempfile import TemporaryDirectory
from klygo import config, files

# Run this cell again to replace only this notebook's temporary workspace.
if "tutorial_workspace" in globals():
    tutorial_workspace.cleanup()
tutorial_workspace = TemporaryDirectory(prefix="klygo-config-")
root = files.path(tutorial_workspace.name)
print("Example directory:", root)
```

```python
settings_path = root / "settings.yaml"
created = config.create(settings_path, {"model": {"batch": 32}}, overwrite=True, verbose=False)
print("Loaded batch size:", config.get(created, "model.batch"))
updated = config.set(created, "model.lr", 0.005)
print("Diff:", config.diff(created, updated)["modified"])
```

```python
from klygo import Config

manager = Config(settings_path)
manager.read(verbose=False)
manager.update({"training": {"epochs": 50}})
exported = manager.export_file("resolved", ".toml", output_dir=root, overwrite=True, verbose=False)
print("Exported file:", files.name(exported), "| exists:", files.is_file(exported))
```

```python
if "tutorial_workspace" in globals():
    tutorial_workspace.cleanup()
    del tutorial_workspace
print("Temporary workspace cleaned up.")
```

## Remember

- Import functional helpers with `from klygo import config, files` and the stateful manager with `from klygo import Config`.
- Functional helpers in `mapping.py`, `structure.py`, and `environment.py` (`merge`, `update`, `flatten`, `unflatten`, `from_env`) and `set()` return new objects without mutating your input mapping, whereas `delete()` removes a key in place.
- File writes (`save`, `convert`, `export`, `create`, `Config.export_file`) protect existing files unless `overwrite=True` is passed.
- Continue to [Document_Config_Io.ipynb](Document_Config_Io.ipynb) or explore sibling modules: [`klygo.files`](../files/Document_Files_Overview.ipynb) and [`klygo.archive`](../archive/Document_Archive_Overview.ipynb).
