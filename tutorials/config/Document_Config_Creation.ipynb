## Prepare workspace

```python
from tempfile import TemporaryDirectory
from klygo import config, files

# Run this cell again to replace only this notebook's temporary workspace.
if "tutorial_workspace" in globals():
    tutorial_workspace.cleanup()
tutorial_workspace = TemporaryDirectory(prefix="klygo-config-")
root = files.path(tutorial_workspace.name)
print("Example directory:", root)
```

## Jump to an API

[`defaults`](#defaults) · [`create`](#create)

<a id="defaults"></a>

### `config.defaults`

Return a fresh copy of Klygo's default configuration with optional deep overrides.

**Signature:** `config.defaults(default_data: Mapping[str, Any] | box.Box | None = None) -> dict[str, Any]`

```python
base_defaults = config.defaults()
custom_defaults = config.defaults({"model": {"batch": 64, "name": "yolov8s"}})
base_defaults["model"]["batch"] = 999
print("Fresh defaults remain isolated:", config.defaults()["model"]["batch"])
print("Custom defaults:", custom_defaults)
```

```python
# Error example: non-mapping override
try:
    config.defaults(["invalid"])
except TypeError as exc:
    print("TypeError ->", exc)
```

<a id="create"></a>

### `config.create`

Create and reload a configuration file from defaults.

**Signature:** `config.create(path: str | pathlib.Path, default_data: Mapping[str, Any] | box.Box | None = None, overwrite: bool = False, verbose: bool = True) -> box.Box`

```python
config_path = root / "project.yaml"
created = config.create(config_path, {"model": {"epochs": 25}}, overwrite=True, verbose=False)
print("Created file exists:", files.is_file(config_path))
print("Loaded Box:", created.model.name, "| epochs:", created.model.epochs)
```

```python
# Error example: refusing to overwrite an existing file when overwrite=False
try:
    config.create(config_path, verbose=False)
except FileExistsError as exc:
    print("FileExistsError ->", exc)
```

## Put it together

```python
template = config.defaults({"project": {"name": "demo"}, "training": {"epochs": 25}})
init_file = root / "initialized.yaml"
created_cfg = config.create(init_file, template, overwrite=True, verbose=False)
print("Created file exists:", files.is_file(init_file), "| epochs:", created_cfg.training.epochs)
```
